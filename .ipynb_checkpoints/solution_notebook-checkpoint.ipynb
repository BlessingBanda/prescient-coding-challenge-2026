## 1. Imports

```python
from __future__ import annotations

import numpy as np
import pandas as pd
```

## 2. Parameters

All tuneable values are kept in `PARAMS` as required by the challenge.

```python
PARAMS = {
    "vol_days":    250,     # lookback for the volatility estimate
    "tilt_size":   0.06,    # how far a 1-sigma signal moves a weight
    "trade_speed": 0.10,    # fraction of the gap to yesterday we close per day
}

ACTIVE_BAND = 0.10       # per asset, distance from benchmark
ACTIVE_BUDGET = 0.40     # total, summed over assets
EQUITY = ["SA_EQUITY", "GLOBAL_EQUITY"]
EQUITY_CAP = 0.75        # total equity, whatever the bands allow
GOLD_CAP = 0.10
```

## 3. Build the trading signal

The current implementation uses inverse volatility as the signal. This is the part intended to be replaced/improved.

```python
def build_signal(hist, params) -> pd.Series:
    """Score per asset. Positive means overweight, negative means underweight."""

    vol = hist.returns.tail(int(params["vol_days"])).std() * np.sqrt(252)
    score = (1.0 / vol.replace(0.0, np.nan)).reindex(hist.assets).fillna(0.0)

    # Standardise so the signal scale is stable through time.
    if score.std() > 0:
        score = (score - score.mean()) / score.std()

    return score
```

## 4. Make the portfolio legal

This function enforces the competition's portfolio constraints.

```python
def make_legal(weights: pd.Series, hist) -> pd.Series:
    """Force weights to satisfy every competition rule."""

    bm = hist.benchmark
    active = weights.reindex(hist.assets).astype(float) - bm

    for _ in range(50):
        # Rule 2: per-asset active-weight band.
        active = active.clip(lower=-ACTIVE_BAND, upper=ACTIVE_BAND)

        # Keep weights non-negative.
        active = active.clip(lower=-bm)

        # Rule 4: total equity cap.
        eq_excess = (bm[EQUITY] + active[EQUITY]).sum() - EQUITY_CAP
        eq_full = eq_excess > -1e-12

        if eq_excess > 0:
            floor = np.maximum(-ACTIVE_BAND, -bm[EQUITY])
            down = (active[EQUITY] - floor).clip(lower=0)

            if down.sum() > 1e-15:
                active[EQUITY] = (
                    active[EQUITY]
                    - eq_excess * down / down.sum()
                )

        # Rule 5: gold cap.
        active["GOLD"] = min(
            active["GOLD"],
            GOLD_CAP - bm["GOLD"]
        )

        # Portfolio weights must sum to 1.
        excess = active.sum()

        if abs(excess) < 1e-12:
            break

        # Give the correction to assets that can absorb it.
        room = (
            ACTIVE_BAND - active
            if excess < 0
            else (active + bm).clip(lower=0)
        )
        room = room.clip(lower=0)

        if excess < 0 and eq_full:
            room[EQUITY] = 0.0

        if room.sum() <= 1e-15:
            break

        active = active - excess * room / room.sum()

    # Rule 3: total active weight <= 40%.
    total = active.abs().sum()

    if total > ACTIVE_BUDGET:
        active = active * (ACTIVE_BUDGET / total)

    return bm + active
```

## 5. Generate daily portfolio weights

The harness calls this function once per trading day.

```python
def generate_weights(hist, prev_weights, params):
    """Return the six portfolio weights to hold on hist.date."""

    bm = hist.benchmark

    # Not enough history to estimate anything: use the benchmark.
    if len(hist.returns) < 260:
        return bm.to_dict()

    # 1. Build the signal and create target weights around the benchmark.
    signal = build_signal(hist, params)
    target = make_legal(
        bm + float(params["tilt_size"]) * signal,
        hist
    )

    # 2. Trade gradually toward the target.
    prev = prev_weights.reindex(hist.assets)
    w = prev + float(params["trade_speed"]) * (target - prev)

    return make_legal(w, hist).to_dict()
```

## 6. Quick reference

The competition harness expects:

`generate_weights(hist, prev_weights, params) -> weights`

The six returned weights must sum to 1 and satisfy the competition constraints.
